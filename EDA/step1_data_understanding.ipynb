# Step 1: 데이터 이해 및 품질 점검

## 목표
- 각 데이터 파일의 기본 구조 파악
- 결측치, 중복 확인
- 기본 통계량 파악 (유저 수, 아이템 수, 분포 등)
- 메타데이터 커버리지 확인
- ID 연속성/범위 확인

```python
import pandas as pd
import numpy as np
import json
import os
from datetime import datetime

# 경로 설정
DATA_PATH = '../data/train/'

print("pandas version:", pd.__version__)
print("numpy version:", np.__version__)
```

## 1. 데이터 로드 및 기본 정보 확인

```python
# 1-1. train_ratings.csv 로드
ratings = pd.read_csv(os.path.join(DATA_PATH, 'train_ratings.csv'))
print("=" * 60)
print("[train_ratings.csv]")
print(f"Shape: {ratings.shape}")
print(f"Memory Usage: {ratings.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
print(f"\nDtypes:\n{ratings.dtypes}")
print(f"\nHead:\n{ratings.head(10)}")
print(f"\nTail:\n{ratings.tail(5)}")
```

```python
# 1-2. titles.tsv 로드
titles = pd.read_csv(os.path.join(DATA_PATH, 'titles.tsv'), sep='\t')
print("=" * 60)
print("[titles.tsv]")
print(f"Shape: {titles.shape}")
print(f"\nDtypes:\n{titles.dtypes}")
print(f"\nHead:\n{titles.head(10)}")
```

```python
# 1-3. years.tsv 로드
years = pd.read_csv(os.path.join(DATA_PATH, 'years.tsv'), sep='\t')
print("=" * 60)
print("[years.tsv]")
print(f"Shape: {years.shape}")
print(f"\nDtypes:\n{years.dtypes}")
print(f"\nHead:\n{years.head(10)}")
print(f"\nYear range: {years['year'].min()} ~ {years['year'].max()}")
```

```python
# 1-4. genres.tsv 로드
genres = pd.read_csv(os.path.join(DATA_PATH, 'genres.tsv'), sep='\t')
print("=" * 60)
print("[genres.tsv]")
print(f"Shape: {genres.shape}")
print(f"\nDtypes:\n{genres.dtypes}")
print(f"\nHead:\n{genres.head(10)}")
print(f"\nUnique genres: {genres['genre'].nunique()}")
print(f"Genre list: {sorted(genres['genre'].unique())}")
```

```python
# 1-5. directors.tsv 로드
directors = pd.read_csv(os.path.join(DATA_PATH, 'directors.tsv'), sep='\t')
print("=" * 60)
print("[directors.tsv]")
print(f"Shape: {directors.shape}")
print(f"\nDtypes:\n{directors.dtypes}")
print(f"\nHead:\n{directors.head(10)}")
print(f"\nUnique directors: {directors['director'].nunique()}")
```

```python
# 1-6. writers.tsv 로드
writers = pd.read_csv(os.path.join(DATA_PATH, 'writers.tsv'), sep='\t')
print("=" * 60)
print("[writers.tsv]")
print(f"Shape: {writers.shape}")
print(f"\nDtypes:\n{writers.dtypes}")
print(f"\nHead:\n{writers.head(10)}")
print(f"\nUnique writers: {writers['writer'].nunique()}")
```

```python
# 1-7. Ml_item2attributes.json 로드
with open(os.path.join(DATA_PATH, 'Ml_item2attributes.json'), 'r') as f:
    item2attr = json.load(f)

print("=" * 60)
print("[Ml_item2attributes.json]")
print(f"Number of items: {len(item2attr)}")
print(f"\nSample (first 5 items):")
for i, (k, v) in enumerate(item2attr.items()):
    if i >= 5:
        break
    print(f"  item {k}: {v}")
```

## 2. 결측치 및 중복 확인

```python
# 2-1. 각 데이터프레임 결측치 확인
print("=" * 60)
print("[결측치 현황]")
print("\n--- train_ratings ---")
print(ratings.isnull().sum())

print("\n--- titles ---")
print(titles.isnull().sum())

print("\n--- years ---")
print(years.isnull().sum())

print("\n--- genres ---")
print(genres.isnull().sum())

print("\n--- directors ---")
print(directors.isnull().sum())

print("\n--- writers ---")
print(writers.isnull().sum())
```

```python
# 2-2. train_ratings에서 (user, item) 중복 확인
print("=" * 60)
print("[train_ratings 중복 확인]")
dup_count = ratings.duplicated(subset=['user', 'item']).sum()
print(f"(user, item) 중복 행 수: {dup_count}")

if dup_count > 0:
    print("\n중복 샘플:")
    dup_mask = ratings.duplicated(subset=['user', 'item'], keep=False)
    print(ratings[dup_mask].sort_values(['user', 'item']).head(20))
```

```python
# 2-3. 메타데이터 테이블에서 item 중복 확인
print("=" * 60)
print("[메타데이터 item 중복 확인]")
print(f"titles - item 중복: {titles['item'].duplicated().sum()} (unique: {titles['item'].nunique()})")
print(f"years - item 중복: {years['item'].duplicated().sum()} (unique: {years['item'].nunique()})")
print(f"genres - item 중복: {genres['item'].duplicated().sum()} (unique: {genres['item'].nunique()}) [다대다 관계로 중복 허용]")
print(f"directors - item 중복: {directors['item'].duplicated().sum()} (unique: {directors['item'].nunique()}) [다대다 관계로 중복 허용]")
print(f"writers - item 중복: {writers['item'].duplicated().sum()} (unique: {writers['item'].nunique()}) [다대다 관계로 중복 허용]")
```

## 3. 기본 통계량 파악

```python
# 3-1. 유저/아이템 수
print("=" * 60)
print("[기본 통계]")
n_users = ratings['user'].nunique()
n_items = ratings['item'].nunique()
n_interactions = len(ratings)

print(f"총 유저 수: {n_users:,}")
print(f"총 아이템 수: {n_items:,}")
print(f"총 상호작용 수: {n_interactions:,}")
print(f"Sparsity: {(1 - n_interactions / (n_users * n_items)) * 100:.4f}%")
```

```python
# 3-2. 유저당 시청 아이템 수 분포
user_item_counts = ratings.groupby('user')['item'].count()

print("=" * 60)
print("[유저당 시청 아이템 수 분포]")
print(f"Mean: {user_item_counts.mean():.2f}")
print(f"Median: {user_item_counts.median():.2f}")
print(f"Std: {user_item_counts.std():.2f}")
print(f"Min: {user_item_counts.min()}")
print(f"Max: {user_item_counts.max()}")
print(f"\nPercentiles:")
for p in [1, 5, 10, 25, 50, 75, 90, 95, 99]:
    print(f"  {p}%: {user_item_counts.quantile(p/100):.0f}")
```

```python
# 3-3. 아이템당 시청 유저 수 분포
item_user_counts = ratings.groupby('item')['user'].count()

print("=" * 60)
print("[아이템당 시청 유저 수 분포]")
print(f"Mean: {item_user_counts.mean():.2f}")
print(f"Median: {item_user_counts.median():.2f}")
print(f"Std: {item_user_counts.std():.2f}")
print(f"Min: {item_user_counts.min()}")
print(f"Max: {item_user_counts.max()}")
print(f"\nPercentiles:")
for p in [1, 5, 10, 25, 50, 75, 90, 95, 99]:
    print(f"  {p}%: {item_user_counts.quantile(p/100):.0f}")
```

```python
# 3-4. Timestamp 범위 확인
print("=" * 60)
print("[Timestamp 범위]")
min_ts = ratings['time'].min()
max_ts = ratings['time'].max()

print(f"Min timestamp: {min_ts} ({datetime.fromtimestamp(min_ts)})")
print(f"Max timestamp: {max_ts} ({datetime.fromtimestamp(max_ts)})")
print(f"기간: {(max_ts - min_ts) / (60*60*24):.1f} 일")
```

```python
# 3-5. 인기 아이템 Top 20
print("=" * 60)
print("[인기 아이템 Top 20]")
top_items = item_user_counts.sort_values(ascending=False).head(20)
top_items_df = pd.DataFrame({
    'item': top_items.index,
    'count': top_items.values
})
top_items_df = top_items_df.merge(titles, on='item', how='left')
print(top_items_df.to_string(index=False))
```

## 4. 메타데이터 커버리지 확인

```python
# 4-1. ratings에 등장하는 item 중 메타데이터 없는 경우 확인
rating_items = set(ratings['item'].unique())
title_items = set(titles['item'].unique())
year_items = set(years['item'].unique())
genre_items = set(genres['item'].unique())
director_items = set(directors['item'].unique())
writer_items = set(writers['item'].unique())
item2attr_items = set(int(k) for k in item2attr.keys())

print("=" * 60)
print("[메타데이터 커버리지]")
print(f"\nratings에 등장하는 총 아이템 수: {len(rating_items):,}")
print(f"\n각 메타데이터별 커버리지:")
print(f"  titles: {len(rating_items & title_items):,} / {len(rating_items):,} ({len(rating_items & title_items)/len(rating_items)*100:.2f}%)")
print(f"  years: {len(rating_items & year_items):,} / {len(rating_items):,} ({len(rating_items & year_items)/len(rating_items)*100:.2f}%)")
print(f"  genres: {len(rating_items & genre_items):,} / {len(rating_items):,} ({len(rating_items & genre_items)/len(rating_items)*100:.2f}%)")
print(f"  directors: {len(rating_items & director_items):,} / {len(rating_items):,} ({len(rating_items & director_items)/len(rating_items)*100:.2f}%)")
print(f"  writers: {len(rating_items & writer_items):,} / {len(rating_items):,} ({len(rating_items & writer_items)/len(rating_items)*100:.2f}%)")
print(f"  item2attributes: {len(rating_items & item2attr_items):,} / {len(rating_items):,} ({len(rating_items & item2attr_items)/len(rating_items)*100:.2f}%)")
```

```python
# 4-2. 메타데이터가 없는 아이템 샘플 확인
print("=" * 60)
print("[메타데이터 누락 아이템]")

missing_title = rating_items - title_items
missing_year = rating_items - year_items
missing_genre = rating_items - genre_items
missing_director = rating_items - director_items
missing_writer = rating_items - writer_items

print(f"\ntitle 누락: {len(missing_title)}개")
if len(missing_title) > 0:
    print(f"  샘플: {list(missing_title)[:10]}")

print(f"\nyear 누락: {len(missing_year)}개")
if len(missing_year) > 0:
    print(f"  샘플: {list(missing_year)[:10]}")

print(f"\ngenre 누락: {len(missing_genre)}개")
if len(missing_genre) > 0:
    print(f"  샘플: {list(missing_genre)[:10]}")

print(f"\ndirector 누락: {len(missing_director)}개")
if len(missing_director) > 0:
    print(f"  샘플: {list(missing_director)[:10]}")

print(f"\nwriter 누락: {len(missing_writer)}개")
if len(missing_writer) > 0:
    print(f"  샘플: {list(missing_writer)[:10]}")
```

```python
# 4-3. 메타데이터 누락 아이템의 인기도 확인
print("=" * 60)
print("[메타데이터 누락 아이템의 상호작용 수]")

for name, missing_set in [('title', missing_title), ('year', missing_year), 
                          ('genre', missing_genre), ('director', missing_director),
                          ('writer', missing_writer)]:
    if len(missing_set) > 0:
        missing_interactions = ratings[ratings['item'].isin(missing_set)].shape[0]
        print(f"{name} 누락 아이템의 총 상호작용 수: {missing_interactions:,} ({missing_interactions/len(ratings)*100:.4f}%)")
```

## 5. ID 연속성 및 범위 확인

```python
# 5-1. User ID 분포
print("=" * 60)
print("[User ID 범위]")
user_ids = ratings['user'].unique()
print(f"Min: {user_ids.min()}")
print(f"Max: {user_ids.max()}")
print(f"Unique count: {len(user_ids)}")
print(f"연속적인 경우 예상 개수: {user_ids.max() - user_ids.min() + 1}")
print(f"누락된 ID 수: {(user_ids.max() - user_ids.min() + 1) - len(user_ids)}")
```

```python
# 5-2. Item ID 분포
print("=" * 60)
print("[Item ID 범위]")
item_ids = ratings['item'].unique()
print(f"Min: {item_ids.min()}")
print(f"Max: {item_ids.max()}")
print(f"Unique count: {len(item_ids)}")
print(f"연속적인 경우 예상 개수: {item_ids.max() - item_ids.min() + 1}")
print(f"누락된 ID 수: {(item_ids.max() - item_ids.min() + 1) - len(item_ids)}")
```

```python
# 5-3. Item ID sparse 정도 확인
print("=" * 60)
print("[Item ID Sparse 분석]")
item_ids_sorted = np.sort(item_ids)
gaps = np.diff(item_ids_sorted)
print(f"ID 간격 통계:")
print(f"  Mean gap: {gaps.mean():.2f}")
print(f"  Max gap: {gaps.max()}")
print(f"  Gap > 100인 경우: {(gaps > 100).sum()}개")
```

## 6. 요약

```python
# 전체 요약 출력
print("=" * 60)
print("[Step 1 요약]")
print("=" * 60)
print(f"\n1. 데이터 규모")
print(f"   - 총 유저: {n_users:,}명")
print(f"   - 총 아이템: {n_items:,}개")
print(f"   - 총 상호작용: {n_interactions:,}개")
print(f"   - Sparsity: {(1 - n_interactions / (n_users * n_items)) * 100:.2f}%")

print(f"\n2. 유저당 시청 수")
print(f"   - 평균: {user_item_counts.mean():.1f}개")
print(f"   - 중앙값: {user_item_counts.median():.0f}개")
print(f"   - 범위: {user_item_counts.min()} ~ {user_item_counts.max()}개")

print(f"\n3. 아이템당 시청 유저 수")
print(f"   - 평균: {item_user_counts.mean():.1f}명")
print(f"   - 중앙값: {item_user_counts.median():.0f}명")
print(f"   - 범위: {item_user_counts.min()} ~ {item_user_counts.max()}명")

print(f"\n4. 시간 범위")
print(f"   - 시작: {datetime.fromtimestamp(min_ts)}")
print(f"   - 종료: {datetime.fromtimestamp(max_ts)}")
print(f"   - 기간: {(max_ts - min_ts) / (60*60*24):.0f}일")

print(f"\n5. 메타데이터 커버리지")
print(f"   - titles: {len(rating_items & title_items)/len(rating_items)*100:.1f}%")
print(f"   - years: {len(rating_items & year_items)/len(rating_items)*100:.1f}%")
print(f"   - genres: {len(rating_items & genre_items)/len(rating_items)*100:.1f}%")
print(f"   - directors: {len(rating_items & director_items)/len(rating_items)*100:.1f}%")
print(f"   - writers: {len(rating_items & writer_items)/len(rating_items)*100:.1f}%")

print(f"\n6. ID 특성")
print(f"   - User ID: {user_ids.min()} ~ {user_ids.max()} (sparse: {(user_ids.max() - user_ids.min() + 1) - len(user_ids)}개 누락)")
print(f"   - Item ID: {item_ids.min()} ~ {item_ids.max()} (sparse: {(item_ids.max() - item_ids.min() + 1) - len(item_ids)}개 누락)")
```
